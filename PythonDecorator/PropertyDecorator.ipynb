```python
#Getter Method
class Person:
    def __init__(self, name):
        self.fullname = name 

    @property
    def name(self):# Getter Method
        print("Getting Name")
        return self.fullname.title()


obj = Person("Santosh karki")
obj.name
```

```python
import uuid
class Person:
    def __init__(self, name):
        self.__fullname = name 

    @property
    def user_id(self):# Getter Method
        print("Getting Id")
        return f"{self.__fullname[-5:]}_{str(uuid.uuid4())[:8]}"


obj = Person("Santosh karki")
obj.user_id
```

```python
class Person:
    def __init__(self, name):
        self.__full_name = name

    @property
    def name(self): #<- Getter Method
        print("Getting Name")
        return self.__full_name

    @name.setter
    def name(self, value): #<- Setter Method
        print("setting Name")
        self.__full_name = value

    
```

```python
person = Person("Jack Sparrow")
```

```python
person.name
```

```python
person.name = "Rajesh Hamal"
```

```python
person.name
```

```python
#Deleter Method
class Person:
    def __init__(self, name):
        self.__full_name = name

    @property
    def name(self): #<- Getter Method
        print("Getting Name")
        return self.__full_name

    @name.setter
    def name(self, value): #<- Setter Method
        print("setting Name")
        self.__full_name = value

    @name.deleter
    def name(self):#Deleter Method
        print("Deleting Name")
        #self.__fullname = "" -> option
        del self.__full_name
```

```python
person = Person("Captain America")
person.name
```

```python
person.name = "John Cena"
```

```python
person.name
```

```python
del person.name
```

```python
person.name
```
