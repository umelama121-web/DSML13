```python
#Implementation

def decorator(func):#-> Function as argument/parameter
    #inner function
    def wrapper():
        print("Before function call")
        print(func().upper())#-> actual function call
        print("After function call")

    return wrapper #-> return function as value
```

```python
@decorator
def greet():
    return "Hello world"
```

```python
greet()
```
