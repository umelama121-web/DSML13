Regular Expression in Python

It works within text only

```python
import re
```

```python
string = "Due to Wikipedia's increasing popularity, some editions, including the English version, have introduced editing restrictions for certain cases. "
print(string)
```

```python
pattern = r"increasing"
match = re.search(pattern , string)
```

```python
if match:
    print("Match Starts:", match.start())
    print("Match End:", match.end())
```

```python
string = "Due to Wikipedia's increasing popularity, some editions, including the USA English version, have introduced editing restrictions for certain cases. For instance, on the English Wikipedia and some other language editions, only users with 10 edits that have an account that is four days USA old may create a new article."

pattern = r"[A-Z]{2,}"

```

```python
match = re.findall(pattern, string)
match
```

Replacing

```python
repl = "---"
new_string = re.sub(pattern, repl, string, count=0 )
```

```python
new_string
```

Split

```python


pattern = r"\[\d+\]" #[78] [98]

Segment = re.split(
    pattern,
    "Hello [78] , I am wikipedia[2] ok", 
    maxsplit=0
)
```
