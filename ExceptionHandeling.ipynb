```python
try:
    num = int(input("Enter number"))
    print(6/num)
except ValueError as e:
    print(e)
    print("Invalid Input")
    print(0)
```

```python
try:
    num = int(input("Enter number"))
    print(6/num)
except ValueError as e:
    print(0)

except ZeroDivisionError as e:
    print ("Cannot devide by zero")
    print(0)
except Exception as e: #Don't use often
    print("unknown Error", e)
    print(0)
```

```python
value = "abc"
try:
    print(int(value))
except (ValueError, TypeError) as e:
    print(f"{value} - Cannot convert to int:", e)
```

```python
class InvalidageException(Exception):
    """ Raise warn input age is less than 18"""
    pass

def check_customer(age):
    if age >= 18:
        print("Good to go")
    else:
        raise InvalidageException("Must be an adult to buy")
```

```python
try:
    check_customer(
        int(input("Enter Age"))
    )
except InvalidageException as e:
    print("Taransaction Cancled: customer under age 18")
except (ValueError, TypeError) as e:
    print("Invalid Age - Transaction Calcled")
```
