<!-- PassCheck Class
pass_check = PassCheck("hello23456")
 
pass_check() -> Strong or Weak Password
 
Strong:

* length >= 8

* alpha + numerical
  -->

```python
class PassCheck:
    def __init__(self, password: str):
        self.password = password


    def __str__(self):
        return f"{self.password[0]}***{self.password[-1]}"

    def __call__(self):
        if len(self.password) >= 8 and self.password.isalnum():
            return "Strong Password"
        return "Weak Password"

pass_check = PassCheck("hello23456")
pass_check()
```
