```python
def calculate_std(*numbers :int | float):
    N = len(numbers)
    mean = sum(numbers) / N

    numerator = sum((number - mean)** 2 for number in numbers)
    std = (numerator / N) ** 0.5

    return round(std, 2)
```

```python
calculate_std(3, 5, 67, 8, 23, 10)
```

```python
# Calculating Student Results:
# Develop a class to accept a student's name and marks in three subjects, then calculate and display the total and average marks.

# total() , average() , display()
```

```python
class Student:

    def __init__(self, name: str, *marks: int):
        self.name = name
        self.marks = marks

    def total(self):
        return sum(self.marks)

    def average(self):
        total = self.total
        return self.total() / len(self.marks)

    def display(self):
        print(f"Name: {self.name}")
        print(f"Total Marks : {self.total()}")
        print(f"Average Marks: {self.average(): .2f}")

umesh = Student("Umesh", 75, 89, 90)
umesh.display()
```

```python
# Designing a simple hierarchy of employees in a company
 
#     - The Person class stores basic details like name and age.
 
#     - The Employee class inherits from Person and adds an employee ID.
 
#     - The Manager class inherits from Employee and adds a department.
 
# Write these classes and implement a method get_details() in the Manager class that returns a string with name, age, employee ID, and department.
 
```

```python
class Person:
    def __init__(self, name:str, age:int):
        self.name = name
        self.age = age

class Employee(Person):
    def __init__(self, name:str, age:int, empl_id: int):
        super().__init__(name, age)
        self.empl_id= empl_id
class Manager(Employee):
    def __init__(self, name:str, age:str, empl_id:int, department:str):
        super().__init__(name, age, empl_id)
        self.department = department

    def get_details(self):
        print(f"Name : {self.name}, \n Age : {self.age}, \n Employe Id : {self.empl_id}, \n Department : {self.department}")

manager = Manager("Umesh Tamang", 27, 12, "Data Science")
manager.get_details()
```

```python
# ## Create a Pet class with hidden energy stats:
# * Private Attribute: __energy (integer, default 50).
# * Attributes: name (string), children (list of Pet objects, defaults to empty list).
# * Methods:
#     * \_\_init\_\_(self, name, energy=50): Initializes name, sets initial __energy, and creates an empty children list.
#     * get_energy(): Returns the private __energy value.
#     * set_energy(value): Updates __energy, but enforcing a range of 0 to 100:
#         * If value > 100, set __energy = 100.
#         * If value < 0, set __energy = 0.
#     * add_child(child_pet): Appends another Pet instance to self. Children.

class Pet:
    def __init__(self, name, energy = 50):
        self.name = name
        self.__energy = 0
        self.children = []

        self.set_energy(energy)


    def get_energy(self):
        return self.__energy 

    def set_energy(self, value):
        if value > 100:
            self.__energy = 100
        elif value < 0:
            self.__energy = 0
        else:
            self.__energy = value

    def add_child(self, child_pet):
        self.children.append(child_pet)

pet1 = Pet("lucy", energy = 50)
```

```python
# Create a RoboPet class that inherits from Pet:
# Attributes: Inherits name, __energy, and children.
# Override Method:
# Override set_energy(value) so that robot pets get a 20% bonus to any energy set (e.g., setting energy to 50 actually gives 60 energy). It must still follow the max cap of 100.
```

```python
class RoboPet(Pet):

    def set_energy(self, value):
        bonous_energy = value * 1.2  #20%
        super().set_energy(bonous_energy)

grandpa = RoboPet("Cyber-Rex", energy = 50)
grandpa.get_energy()
```

```python
# Write a standalone function get_total_family_energy(pet):
# Base Case: If pet has no children, return its own energy (pet.get_energy()).

# Recursive Case: Return the pet's own energy plus the sum of its children energy.
```

```python
def get_total_family_energy(pet):
    total = pet.get_energy()

    for child in pet.children:
        total += get_total_family_energy(child)

    return total

get_total_family_energy(grandpa)
```
