```python
def calculate_std(*numbers :int | float):
    N = len(numbers)
    mean = sum(numbers) / N

    numerator = sum((number - mean)** 2 for number in numbers)
    std = (numerator / N) ** 0.5

    return round(std, 2)
```

```python
calculate_std(3, 5, 67, 8, 23, 10)
```

```python
#Recursive Function
###Factorial of 5
def factorial(n):
    if n in [0, 1]:
        return 1
    else:
        return factorial(n-1) * n
```

```python
# Calculating Student Results:
# Develop a class to accept a student's name and marks in three subjects, then calculate and display the total and average marks.

# total() , average() , display()
```

```python
class Student:

    def __init__(self, name: str, *marks: int):
        self.name = name
        self.marks = marks

    def total(self):
        return sum(self.marks)

    def average(self):
        total = self.total
        return self.total() / len(self.marks)

    def display(self):
        print(f"Name: {self.name}")
        print(f"Total Marks : {self.total()}")
        print(f"Average Marks: {self.average(): .2f}")

umesh = Student("Umesh", 75, 89, 90)
umesh.display()
```

```python
# Designing a simple hierarchy of employees in a company
 
#     - The Person class stores basic details like name and age.
 
#     - The Employee class inherits from Person and adds an employee ID.
 
#     - The Manager class inherits from Employee and adds a department.
 
# Write these classes and implement a method get_details() in the Manager class that returns a string with name, age, employee ID, and department.
 
```

```python
class Person:
    def __init__(self, name:str, age:int):
        self.name = name
        self.age = age

class Employee(Person):
    def __init__(self, name:str, age:int, empl_id: int):
        super().__init__(name, age)
        self.empl_id= empl_id
class Manager(Employee):
    def __init__(self, name:str, age:str, empl_id:int, department:str):
        super().__init__(name, age, empl_id)
        self.department = department

    def get_details(self):
        print(f"Name : {self.name}, \n Age : {self.age}, \n Employe Id : {self.empl_id}, \n Department : {self.department}")

manager = Manager("Umesh Tamang", 27, 12, "Data Science")
manager.get_details()
```
