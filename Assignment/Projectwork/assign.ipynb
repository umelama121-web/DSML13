Supermarket receipt

```python
item = input("Name of the items: ")
price = float(input("price of the items: "))
quentity = int(input("quentity of the items: "))

total = price  * quentity
vat = total * 0.13
subtotal = total + vat


print(f"Item: {item} \n Price: {price} \n Quantity: {quentity}")
print(f"total: {total} \n Vat: {vat}  \n Subtotal: {subtotal}")




```
