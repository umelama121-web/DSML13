Loop over Tuples

```python
prices = {"tea": 20, "coffee": 50, "momo": 150}

for item in prices:                   # keys only
    print(item)

for value in prices.values():          # values only
    print(value)

for item, value in prices.items():    # both keys and values
    print(f"{item}: {value}")
```

```python
marks = [45, 80, 32, 67, 25]
passed = 0
failed = 0

for m in marks:
    if m >= 40:
        print(f"{m}: Pass")
        passed += 1
        print(f"Total passed students: {passed}")
    elif m < 40:
        failed += 1
        print(f"Total failed students: {failed}")
    else:
        print(f"{m}: Fail")

print(f"{passed} students passed")
```

While Looop

```python
count = 1

while count <= 5:
    print(count)
    count += 1

print("Done!")
```

```python
tasks = []

task = input("Add a task (or 'quit'): ")

while task != "quit":
    tasks.append(task)
    task = input("Add a task (or 'quit'): ")

print("Your tasks:", tasks)
```

Loop inside loop [Nasted loop]

```python
for row in range(1, 4):
    for col in range(1, 4):
        print(f"{row} x {col} = {row * col}")
    print("---")
```

```python
for row in range(1, 4):
    for col in range(1, 6):
        if col ==2 and row == 2:
            continue
        print(f"{row} x {col} = {row * col}")
    print("---")
```

```python
# Nested loop
for day in range(1,4):
    print(f"Day {day}")
    for meal in range(1,4):
        print (f"Meeal {meal}")
```

```python
for row in range(1, 5):
    for star in range(row):
        print("*", end=" ")
    print()
```
