Calculator


```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

operator = input("choose an operator (+, -, *, / , %): ")

match operator:
    case "+":
        print(num1 + num2)
    case "-":
        print(num1 - num2)
    case "*":
        print(num1 * num2)
    case "/":
        print(num1 / num2)
    case "%":
        print(num1 % num2)
    case _:
        print("Invalid operator")

```
