Keys of Dictionary

```python
prices ={"tea": 20, "coffe" : 50, "milk": 30}

print(prices.keys())
print(prices.values())
print(prices.items())

print(list(prices))
print(list(prices.values()))
```

```python
print("tea" in prices)
print(50 in prices)
print(50 in prices.values())
```

```python
print(len(prices))
print(len(prices.values()))
```

```python
print(sum(prices.values())) #Work only in int.
```

Joining two dictionary

```python
a = {"tea": 30, "milk": 20}
b = {"milk": 20, "juice": 30}
print(a|b)
```

Setdefult() and Formskeys()

```python
s = {"name": "Ram"}
print(s.setdefault("age",20))
print(s)
```

```python
print(s.setdefault("name", "sita"))
print(s)
```

```python
marks = dict.fromkeys(["math", "science", "english"])
print(marks)
```

Zip and covering

```python
names = ["Ram", "Sita", "Hari"]
marks = [20, 30, 40]
dict1 = dict(zip(names,marks))
print(dict1)
```

Dictionary inside dictionary

```python
school = {
    "ram" : {"age": 20, "city": "Ktm"},
    "sita": {"age" : 19, "city":"Pokhara"}
}
print(school["ram"])
print(school["sita"]["age"])
```

```python
school["ram"]["city"] = "lalitpur"
print(school["ram"])
```
