Typecastig

```python
string = "100"
number = int(string)
print(type(number))  # Output: <class 'int'>
```

```python
num = 123
str = str(num)
print(type(str))  # Output: <class 'str'>
```

```python
num = 1
print(type(num))  # Output: <class 'int'>
bool1 = bool(num)
print(bool1)  # Output: True
print(type(bool1))  # Output: <class 'bool'>
```
