Numbers and strings

```python
name = "Ram"
age = 25

print(f"My name is {name} and I am {age} years old.")
```

```python
number = 12
number2 = 1.5
num1 = number
num2 = number2

add = num1 + num2
sub = num1 - num2
mul = num1 * num2
div = num1 / num2
print(f"Addition: {num1} + {num2} = {add}")
print(f"Subtraction: {num1} - {num2} = {sub}")
print(f"Multiplication: {num1} * {num2} = {mul}")
print(f"Division: {num1} / {num2} = {div}")

power = num1 ** num2
print(f"Power: {num1} ** {num2} = {power}")
```

```python
price = 200
queantity = 3
discount = 75
tax = 0.13

total = (price * queantity) - discount
total_tax = total * tax
print(f"total amount with tax: {total + total_tax}")

```

```python
num1 = int(input("Enter first number: "))
new_num1 = num1 + 10
print(num1)
print(new_num1)
```

```python
score = 85
price = 200
discount = 5

score += 5
price -= 10
discount *= 2
print(f"Updated score: {score}")
print(f"Updated price: {price}")
print(f"Updated discount: {discount}")
```

Strings

```python
word = "Good morning"

word1 = word.upper()
word2 = word.lower()
woerd3 = word.title()

print(word1)
print(word2)
print(woerd3)

length = len(word)
print(f"Length of the word '{word}' is: {length}")
```

```python
word = "    Good morning    "

word3 = word.strip()
print(f"Original word: '{word}'")
print(f"Word after stripping: '{word3}'")
```

```python
a = 10
b = 3
c = "ram"
d = True


print(f"Type of a: {type(a)}")
print(f"Type of b: {type(b)}")
print(f"Type of c: {type(c)}")
print(f"Type of d: {type(d)}")
```
