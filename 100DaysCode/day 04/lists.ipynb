```python
score = [84, 92, 76, 88, 95]

print(len(score))  # Output: 5
print(max(score))  # Output: 95
print(min(score))  # Output: 76
print(sum(score))  # Output: 435

score.sort()
print(score[::-1])  # Output: [76, 84, 88, 92, 95]
```

```python
names = ["Alice", "Bob", "Charlie", "David", "Eve"]

names.sort()
print(names)

names.sort(reverse=True)
print(names)
names.sort(key=str.lower)
print(names)
```

```python
# 02_shopping_cart.py

cart = [
    ["Bread", 60, 2],
    ["Milk", 120, 1],
    ["Eggs", 15, 12],
]

grand_total = 0
for item in cart:
    name, price, qty = item
    line_total = price * qty
    print(f"{name}: Rs. {line_total}")
    # "cm"># Your job: add line_total onto grand_total


grand_total += line_total
print(f"Grand total: Rs. {grand_total}")
```
