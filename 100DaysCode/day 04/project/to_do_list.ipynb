```python
task = []

tasks = []

task = input("Add a task (or 'quit'): ")
while task != "quit":
    tasks.append(task)
    task = input("Add a task (or 'quit'): ")

count = 1
for t in tasks:
    print(f"{count}. {t}")
    count += 1

remove = int(input("Enter the task number to remove: "))

removed_task = tasks.pop(remove - 1)
print(f"Removed: {removed_task}")

print("\nUpdated tasks:")
count = 1
for t in tasks:
    print(f"{count}. {t}")
    count += 1
```
