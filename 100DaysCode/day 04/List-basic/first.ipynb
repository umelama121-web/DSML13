Creating a list


```python
fruits = ["apple", "banana", "orange"]

print(fruits)
print(fruits[0])  # Accessing the first element
print(fruits[1])  # Accessing the second element
print(len(fruits))  # Getting the length of the list
```

Slicing - grabbing a chunk

```python

nums = [10, 20, 30, 40, 50, 60]

nums[1:4]     # [20, 30, 40]
nums[:3]      # [10, 20, 30]  — from the start
nums[3:]      # [40, 50, 60]  — to the end
nums[::2]     # [10, 30, 50]  — every 2nd item
nums[1:6:2]   # [20, 40, 60]  — start, stop, AND step together
nums[::-1]    # [60, 50, 40, 30, 20, 10]  — reversed!
```

```python
fruits = ["apple", "banana", "orange"]
fruits[1]
fruits[0:2]
fruits[0:2] = ["kiwi" , "mango"]  # This line has a syntax error; it should be fruits[0:2] = ["kiwi", "mango"]
print(fruits)  # Output: ['kiwi', 'mango', 'orange']
```
