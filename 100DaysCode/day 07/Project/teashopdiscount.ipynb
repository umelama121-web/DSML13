```python
menu = {"tea": 20, "coffee": 50, "momo": 150}
print("Menu:", menu)

item = input("What do you want? ").strip().lower()

if item in menu:
    qty = int(input("How many? "))
    total = menu[item] * qty
    print(f"Total: Rs. {total}")

    if total >= 500:
        with_dis = total * 0.1
        all_total = total - with_dis
        print(f"with 10 % discount: {all_total: .2f}")
    elif total >=200:
        with_dis = total * 0.05
        all_total = total - with_dis
        print(f"With 5% discount: {all_total: .2f} ")
    else:
        print("NO discount")
        print(f"Your bill : {total}")
else:
    print("Sorry, we don't have that")

# Your job (inside the if block):
# 500 or more -> 10% off, 200 or more -> 5% off,
# else no discount. Print the final bill.
```
