```python
day = input("Day: ")

# with if / elif
if day == "sat":
    print("Holiday")
elif day == "fri":
    print("Half day")
else:
    print("School day")

# with match-case (Day 03)
match day:
    case "sat":
        print("Holiday")
    case "fri":
        print("Half day")
    case _:
        print("School day")
```
