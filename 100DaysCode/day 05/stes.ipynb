```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6, 7}
print(set1 | set2)
print(set1 & set2)
print(set1 - set2)
print(set2 - set1)
print(set1 ^ set2)
```

```python
small = {1, 2 }
big = {1, 2, 3, 4}

print(small >= big)
print(big.issubset(small))
print(big >= small)
print(big.issuperset(small))
```

Copy

```python
a = {1, 2, 3,}
b = a
b.add(4)
print(b)
print(a)

c = a.copy()
c.add(5)
print(c)
print(a)
```
