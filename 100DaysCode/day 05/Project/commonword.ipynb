```python

s1 = input("Sentence 1: ").lower()
s2 = input("Sentence 2: ").lower()

words1 = set(s1.split())
words2 = set(s2.split())

common = words1 & words2
print("Common words:", sorted(common))

all_word = words1 | words2 
print("all words:", sorted(all_word))

count =  len((common)), len((all_word))
print(count)
```
