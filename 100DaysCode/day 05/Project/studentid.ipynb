Student ID card


```python
name = input("Name:")
age = int(input("Age:"))
city = input("City:")

student = (name, age, city)
n, a, c = student

print("========ID==========")
print("Name:", n)
print("Age:", a)
print("City:", c)
print("=====================")

marks = (70, 85, 90)
maxium_mark = max(marks)
minium_mark = min(marks)
sum_mark = sum(marks)

print(f"Maxium mark: {maxium_mark} \n Minium mark: {minium_mark} \n Sum of marks: {sum_mark}")

student = list(student)
student[2] = input("Change the city")

student = tuple(student)
print(f"Updated Information : {student}")
```
