```python
def factorial(n):
    # Base case
    if n in (1,0):
        return 1

    else:
        return factorial(n-1) * n 

factorial(5)
```

```python
#Memonization

cache = {0: 1, 1: 1}

def factorial(n):
    # Base case
    if n in cache.keys():
        return cache[n]

    else:
        res = factorial(n-1) * n 
        cache[n] = res

        return res

factorial(5)
```
